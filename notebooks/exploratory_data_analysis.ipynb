## 3. Domain Feature Engineering
We introduce `is_obese` ($BMI \ge 30$) and the interaction term `obese_smoker` ($is\_obese \times smoker$). In actuarial science, obesity combined with tobacco smoking presents exponential morbidity risk.

```python
df_engineered = df.copy()
df_engineered['is_obese'] = (df_engineered['bmi'] >= 30.0).astype(int)
df_engineered['obese_smoker'] = df_engineered['is_obese'] * (df_engineered['smoker'] == 'yes').astype(int)
df_engineered.head()
```
